```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

ocen = ['Отлично', 'Хорошо', 'Удовлетворительно', 'Неудовлетворительно']
kolvo = [20, 35, 30, 15]

plt.pie(kolvo, 
        labels=ocen,
        autopct='%1.1f%%',  # процентные подписи с одним знаком после запятой
       )
plt.title('Распределение оценок студентов')
plt.show()
```
