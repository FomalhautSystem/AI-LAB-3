```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import quad

def f(x):
    return 3*x**4 - 2*x**3 + x**2 - 5

a, b = -1, 1
integral_value, error = quad(f, a, b)

print(f"Значение интеграла: {integral_value:.4f}")

x = np.linspace(a - 0.2, b + 0.2, 500)
y = f(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, 'b-', linewidth=2, label=r'$f(x) = 3x^4 - 2x^3 + x^2 - 5$')

x_fill = np.linspace(a, b, 200)
y_fill = f(x_fill)
plt.fill_between(x_fill, y_fill, alpha=0.3, color='skyblue', label=f'Площадь = {integral_value:.3f}')

plt.axhline(0, color='black', linestyle='--')
plt.axvline(a, color='red', linestyle='--')
plt.axvline(b, color='red', linestyle='--')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r'Площадь под кривой $f(x)$ на $[-1, 1]$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
