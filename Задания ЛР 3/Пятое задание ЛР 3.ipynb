```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

ocen = ['Отлично', 'Хорошо', 'Удовлетворительно', 'Неудовлетворительно']
kolvo = [20, 35, 30, 15]

plt.figure(figsize=(12, 8))
plt.bar(ocen, kolvo, edgecolor='black')

plt.title('Распределение оценок студентов')
plt.xlabel('Оценка')
plt.ylabel('Количество студентов')

for i, v in enumerate(kolvo):
    plt.text(i, v + 1, str(v), ha='center')

plt.grid()
plt.show()
```
