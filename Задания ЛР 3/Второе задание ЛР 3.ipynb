```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

x = np.linspace(-1, 1, 100)

y1 = x           
y2 = x ** 2      
y3 = x ** 3      

plt.plot(x, y1, label='$y = x$')         
plt.plot(x, y2, label='$y = x^2$')         
plt.plot(x, y3, label='$y = x^3$')        

plt.title('Сравнение степенных функций')

plt.xlabel('x')
plt.ylabel('y')

plt.grid(True)
plt.legend()
plt.axis('equal')

plt.show()
```
