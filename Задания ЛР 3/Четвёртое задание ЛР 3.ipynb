```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

data = np.random.randn(1000)

plt.figure(figsize=(10, 6))
plt.hist(data, bins=30)

mean = np.mean(data)
plt.axvline(mean, color='red', linewidth=2, label=f'Среднее = {mean:.3f}')

plt.xlabel('Значение')
plt.ylabel('Частота')
plt.title('Гистограмма нормального распределения N(0, 1)')
plt.legend()
plt.grid()
plt.show()
```
