```python
import matplotlib.pyplot as plt
%matplotlib inline

countries = ['США', 'Германия', 'Франция', 'Япония', 'Россия']
years = [4, 3.5, 3, 4.5, 4]

plt.barh(countries, years)

plt.title('Среднее время обучения в университетах')
plt.xlabel('Лет обучения')
plt.ylabel('Страны')
plt.grid()
plt.show()
```
